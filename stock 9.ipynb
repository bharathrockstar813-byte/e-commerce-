```python
#converting data column
df["Date"]=pd.to_datetime(df["Date"])
df.info()
```

```python
#sort the values
dfd=df.sort_values("Date")
dfd
```

```python
#set date as index
dfd.set_index("Date", inplace=True)
dfd
```

```python
#so we can create missing date too
dfd=dfd.asfreq("D")
dfd
```

```python
dfd["Stock_Price"]=dfd["Stock_Price"].interpolate()
dfd["Volume"]=dfd["Volume"].interpolate()
dfd
```

```python
#forwared replace a missing value with the previous vale
#backwardpill repalce a missing value with next availlble valuefill
```

```python
#convert data features
dfd["Year"]=dfd.index.year
dfd["Monnth"]=dfd.index.month
dfd["Day"]=dfd.index.day
dfd
dfd["dayofweek"]=dfd.index.dayofweek
dfd
```

```python
dfd["weekend"]=(dfd.index.dayofweek>=5).astype(int)
dfd
```

```python
#lag feature
#a lag feature means use part valuws to predict future value
dfd["lag1"]=dfd["Stock_Price"].shift(1)
dfd
dfd["lag2"]=dfd["Stock_Price"].shift(2)
dfd["moving_average"]=dfd["Stock_Price"].rolling(3).mean()
dfd
```
