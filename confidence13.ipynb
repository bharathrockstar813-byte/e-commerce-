```python
import pandas as pd
df=pd.read_csv("/content/drive/MyDrive/data cleaing/career_dataset.csv")
df
```

```python
print("FIRST 5 RECORDS")
print(df.head())

print("\nLAST 5 RECORDS")
print(df.tail())

print("\nSHAPE (Rows, Columns)")
print(df.shape)

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nDATA TYPES")
print(df.dtypes)
```

```python
#1
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = df.select_dtypes(include=['object']).columns.tolist()

print("\nNUMERICAL COLUMNS")
print(num_cols)

print("\nCATEGORICAL COLUMNS")
print(cat_cols)
```

```python
print("\nMISSING VALUES BEFORE TREATMENT")
print(df.isnull().sum())

# Name → Unknown
df['Name'] = df['Name'].fillna("Unknown")

# Age → Mean
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Gender → Mode
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

# Qualification → Mode
df['Qualification'] = df['Qualification'].fillna(
    df['Qualification'].mode()[0]
)
# Experience_Years → Median
df['Experience_Years'] = df['Experience_Years'].fillna(
    df['Experience_Years'].median()
)

# Expected_Salary → Median
df['Expected_Salary'] = df['Expected_Salary'].fillna(
    df['Expected_Salary'].median()
)

# City → Mode
df['City'] = df['City'].fillna(df['City'].mode()[0])

print("\nMISSING VALUES AFTER TREATMENT")
print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("\nTOTAL DUPLICATES =", duplicate_count)

df = df.drop_duplicates()

print("\nDUPLICATES AFTER REMOVAL")
print(df.duplicated().sum())
```

```python
# Gender Standardization
df['Gender'] = df['Gender'].replace({
    'M': 'Male',
    'F': 'Female'
})

# Column Names to Lowercase
df.columns = df.columns.str.lower()

print("\nUPDATED COLUMN NAMES")
print(df.columns.tolist())
```

```python
numerical_columns = df.select_dtypes(
    include=['int64', 'float64']
).columns

print("\nNUMERICAL COLUMNS")
print(numerical_columns)

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print(f"\nColumn: {col}")
    print("Q1 =", Q1)
    print("Q3 =", Q3)
    print("IQR =", IQR)
    print("Lower Bound =", lower_bound)
    print("Upper Bound =", upper_bound)

    df = df[
        (df[col] >= lower_bound) &
        (df[col] <= upper_bound)
    ]

print("\nOUTLIER REMOVAL COMPLETED")
```

```python
df['applied_date'] = pd.to_datetime(
    df['applied_date'],
    errors='coerce'
)

print("\nDATE COLUMN TYPE")
print(df['applied_date'].dtype)
```
